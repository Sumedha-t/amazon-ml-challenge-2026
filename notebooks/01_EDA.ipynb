```python
gt = ground_truth.copy()

gt["matched_entity_ids"] = gt["matched_entity_ids"].fillna("").astype(str)

gt["n_matches"] = gt["matched_entity_ids"].apply(
    lambda x: 0 if not x.strip() else len(x.split(","))
)

print(gt["n_matches"].value_counts().sort_index())
```

```python
print("Total S1 entities:", len(gt))
print("Singletons (0 matches):", (gt["n_matches"] == 0).sum())
print("Exactly 1 match:", (gt["n_matches"] == 1).sum())
print("Exactly 2 matches:", (gt["n_matches"] == 2).sum())
print("3+ matches:", (gt["n_matches"] >= 3).sum())
print("Maximum matches:", gt["n_matches"].max())
```

```python
match_distribution = (
    gt["n_matches"]
    .value_counts()
    .sort_index()
    .to_frame("count")
)

match_distribution["percentage"] = (
    match_distribution["count"] / len(gt) * 100
)

display(match_distribution)
```

```python
def count_source_matches(match_string):
    if not match_string.strip():
        return pd.Series([0, 0])

    ids = match_string.split(",")

    s2_count = sum(x.startswith("S2-") for x in ids)
    s3_count = sum(x.startswith("S3-") for x in ids)

    return pd.Series([s2_count, s3_count])


gt[["n_s2_matches", "n_s3_matches"]] = gt["matched_entity_ids"].apply(
    count_source_matches
)

print(gt[["n_s2_matches", "n_s3_matches"]].describe())
```

```python
print("\nS2 match count distribution:")
print(gt["n_s2_matches"].value_counts().sort_index())

print("\nS3 match count distribution:")
print(gt["n_s3_matches"].value_counts().sort_index())
```

```python
print("\nS1 entities with at least one S2 match:",
      (gt["n_s2_matches"] > 0).sum())

print("S1 entities with at least one S3 match:",
      (gt["n_s3_matches"] > 0).sum())

print("S1 entities with matches in BOTH:",
      ((gt["n_s2_matches"] > 0) & (gt["n_s3_matches"] > 0)).sum())
```

```python
sample_gt = (
    gt[gt["n_matches"] > 0]
    .sample(10000, random_state=42)
)

pairs = []

for _, row in sample_gt.iterrows():
    s1_id = row["source1_entity_id"]

    for mid in row["matched_entity_ids"].split(","):
        if mid.startswith("S2-"):
            target = s2_lookup.loc[mid]
            source = "S2"
        else:
            target = s3_lookup.loc[mid]
            source = "S3"

        source1 = s1_lookup.loc[s1_id]

        pairs.append({
            "source": source,
            "same_name_exact": source1["business_name"] == target["business_name"],
            "same_address_exact": source1["business_address"] == target["business_address"],
            "same_country": source1["country"] == target["country"],
        })

pair_sample = pd.DataFrame(pairs)

pair_sample.groupby("source")[
    ["same_name_exact", "same_address_exact", "same_country"]
].mean()
```

```python
import re
import html
import unicodedata

def normalize_basic(text):
    if pd.isna(text):
        return ""

    text = html.unescape(str(text))
    text = unicodedata.normalize("NFKC", text)
    text = text.casefold()

    # Remove punctuation/symbols
    text = re.sub(r"[^\w\s]", " ", text, flags=re.UNICODE)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
for col in ["business_name", "business_address"]:
    train_s1[f"{col}_norm"] = train_s1[col].map(normalize_basic)
    train_s2[f"{col}_norm"] = train_s2[col].map(normalize_basic)
    train_s3[f"{col}_norm"] = train_s3[col].map(normalize_basic)
```
